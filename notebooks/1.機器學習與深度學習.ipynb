# 1.機器學習與深度學習
## 關於機器學習 （Machine Learning, ML）的筆記

此記錄機器學習的一點基礎理解


# 機器學習與深度學習的差異是?

機器學習（Machine Learning, ML） 是大範圍的方法論  
深度學習（Deep Learning, DL） 是其中一個特別強、特別熱門的分支  
深度學習 ⊂ 機器學習。 深度學習（Deep Learning）是機器學習（Machine Learning）的一個子集。

# 差異  
1.特徵工程 (Feature Engineering) 的參與度  
這是兩者最本質的差異：  
機器學習（傳統）： 需要人工進行「特徵提取」。  
例如：如果要辨識一張照片是不是貓，工程師需要先寫程式去定義什麼是「尖耳朵」、什麼是「鬍鬚」。機器負責的是根據這些人為定義的特徵來做最終的判斷。

深度學習： 實現了「自動化特徵提取」。  
它模仿人類大腦的神經網路，直接輸入原始數據（如像素），模型會自動在層與層之間學會辨識邊緣、形狀，最後才是複雜的特徵（如貓臉）。  

2.硬體需求與資料量  
機器學習： 在數據量較小時表現穩定，且通常不需要極高規的硬體，普通的 CPU 即可處理大部分的演算法（如線性迴歸、隨機森林）。  

深度學習： 是典型的「大數據驅動」。當數據量達到一定規模後，深度學習的準確度會大幅超越傳統演算法。由於涉及大量的矩陣運算，它高度依賴 GPU 來進行並行計算。  

3.模型結構  
機器學習： 結構通常較為扁平且具備較強的可解釋性（你可以清楚知道為什麼模型做這個決定）。  

深度學習： 結構由多層「人工神經網路」（Artificial Neural Networks）組成。因為層數非常「深」，所以被稱為深度學習。它的缺點是像個「黑盒子」，雖然預測精準，但有時難以解釋內部的邏輯運作過程。  

| 特性 | 機器學習 (傳統) | 深度學習 |
| :--- | :--- | :--- |
| **數據依賴** | 數據量較小時亦可運作良好 | 需要海量數據才能發揮威力 |
| **特徵提取** | 依賴人工專家定義特徵 | 自動從數據中學習特徵 |
| **執行時間** | 訓練時間短，推論速度快 | 訓練時間長（數天或數週） |
| **硬體** | 低成本 CPU 即可 | 高性能 GPU / TPU |
| **底層理論** | 統計學、決策理論 | 神經科學啟發、多層矩陣運算 |

# 深度學習本質上非常數學。

大量依賴：微積分、線性代數、矩陣論  
微積分:尤其：  
梯度下降（Gradient Descent）  
$$ \theta_{new} = \theta - \eta \nabla L(\theta) $$
本質：沿著誤差下降最快方向調整參數。 

機率統計
例如：loss function, likelihood, softmax , Bayesian thinking

最佳化理論
因為本質是：在超高維空間找最佳參數。

#  訓練集、驗證集和測試集 是什麼? 


# 訓練集 (Training Set)  
定義： 這是模型用來學習的數據，包含輸入的特徵（考題）與正確答案（標籤）。  
角色： 【教科書與課後練習】  
運作： 模型會反覆讀取這些數據，調整內部的參數（如權重），試圖找出輸入與輸出之間的數學規律。

# 驗證集 (Validation Set)
定義： 在訓練過程中，模型沒見過、用來初步評估表現的數據。  
角色： 【模擬考】  
運作： 用它來檢查模型有沒有過擬合（Overfitting）——即模型在訓練集考滿分，但遇到沒看過的題目就掛掉。  
我們會根據驗證集的結果來調整「超參數」（例如學習率、網路層數）。注意： 驗證集雖然不參與權重更新，但它會「參與」你對模型的選擇。


這就是：
Gradient Descent  
$$ \theta_{new} = \theta - \eta \nabla L(\theta) $$
其中：
θ = 模型參數  
L(θ) = loss function  
η = learning rate  
模型在訓練集上：不斷降低誤差。



# 測試集 (Test Set)
定義： 模型開發完成、參數全部固定後，最後才拿出來測試的「純淨數據」。  
角色： 【正式大考】  
運作： 它是為了評估模型在「現實世界」遇到完全陌生狀況時的最終能力。在最後一步之前，絕對不能讓模型看到測試集的內容，否則就是「洩題」，會導致準確率失真。  

# 最核心觀念：泛化（Generalization）

AI 真正厲害的不是：記住資料  

而是：沒看過的新資料也有效  

這叫：泛化能力  

# 「過擬合」是什麼 ?
「過擬合（Overfitting）」  :  
過擬合就是模型「記憶」了訓練數據中的雜訊（Noise），而不是「學習」到背後的普遍規律。  

就像一個學生背下了練習題的所有答案（包括題目裡的標點符號），結果考試時考卷稍微改個數字，他就完全寫不出來。  


如果你發現模型在訓練集上的準確率是 $99\%$，但在驗證集上只有 $70\%$，這就是典型的過擬合。  
這時你可能需要：  
簡化模型： 減少神經網路的層數。  
增加數據： 讓模型看更多題目。  
正規化 (Regularization)： 在數學公式中加入懲罰項，不讓模型過度依賴某些特定特徵。

注意：train 一直升。
但 validation：開始下降。  
這通常代表：過擬合開始。  
如果模型在 validation 也表現好，代表學到可泛化規律。

# 除了切資料，還有哪些防過擬合方法？

1.Regularization（正則化） 
限制模型不要太亂。  
$$L = L_0 + \lambda ||w||^2$$
意思：權重太大要罰。避免模型亂震盪。

2.Dropout  
訓練時：隨機關閉部分神經元。  
迫使模型：不要過度依賴某些特徵。  

3.Data Augmentation  
例如圖片：  
旋轉、翻轉、裁切、加噪聲  
讓模型看到：更多變化。  

4.Early Stopping  
看到 validation 開始下降：提前停止。  
不要繼續硬練。  

# 個人理解

1.訓練: 給模型一堆資訊，使之訓練(藉由背後的演算法調整各種參數) 最終完成某個超高維函數的模型   
2.驗證: 給模型有別於訓練集的資料，使之 挑選最佳模型、調整參數  
3.測試:最終模型的評估，給模型前面沒有出現過得資訊，看看表現如何  

因為深度學習本質真的就是：超高維函數逼近。  
  
拿數學來比喻:  
1.訓練:我讀微積分的積分技巧，理解以及做題目做訓練  
2.驗證:我拿其他題目來挑戰自我對積分的理解，若有錯就反思是不是我對積分技巧地不熟練，並優化   
3.測試:我真正去參與積分考試 (無法翻書 無法查找資料) 只能依賴我的理解  

# 「機器學習就是透過資料訓練出一個超高維函數，並透過三層資料過濾機制，確保這個函數不是死記硬背，而是真的具備邏輯推演能力。」

本質上就是利用深度模型的技術，我可以理解為，有個超級強大的黑箱(模型)  因為大多數人類已經難以企及，可能也無法解釋最終結果為什麼函數會長這樣，這邊我稱之為黑箱，而這個黑箱藉由不斷訓練、驗證、調整出一個超高維函數，使出誕生出現代各種超強AI，強大到表現力在各領域上有超越人類的可能  (好比 圍棋  、生圖 等等)

1.黑箱
深度學習（尤其是現在的大語言模型）擁有數十億甚至上兆個參數。這確實是一個超高維函數，它在空間中扭曲、折疊，最終形成了一個極其複雜的「邊界」。  
人類企及： 傳統數學公式（如 $f(x) = ax + b$）我們可以輕易理解。但當函數變成 $f(x_{1 \dots 10^{12}})$，且包含了無數次的非線性轉換時，人類大腦確實無法直觀理解「為什麼這個權重多 $0.0001$ 就能畫出一隻貓」。  
表現力： 這種複雜度給予了它極強的「擬合能力」，讓它能捕捉到人類肉眼或傳統統計學看不見的微小規律。

2.本質：統計模式逼近 (Statistical Pattern Approximation)  
你的理解非常到位。無論 AI 看起來多麼像有「靈魂」，它的底層運作依然是：  

在高維空間中，尋找機率分佈的規律。  

它並不是真的「理解」什麼是圍棋，而是學會了「在這種棋盤分佈下，下在這裡的勝率最高（機率最大）」。它在超高維度下找到了數據的穩定結構，強大到讓人類覺得它具備了直覺。  

3.關於「未見過的開放性問題」 (泛化能力的邊界)  
你舉的「積分開放性問題」案例，精確地點出了目前 AI 的瓶頸：泛化 (Generalization) 與外推 (Extrapolation)。  
內插 (Interpolation) 如果考題在「訓練資料的分佈範圍內」，AI 的表現會強得驚人。  
外推 (Extrapolation)： 如果題目是一個全新的邏輯範式（例如人類歷史上從未出現過的數學領域，或是需要多步邏輯跳躍的開放問題），目前的 AI 高機率會失敗。  

| 你的觀點 | 專業對應 | 評價 |
| :--- | :--- | :--- |
| **超高維函數 / 黑箱** | 神經網路的非線性映射 | 精準。這就是為什麼我們需要對抗訓練來找它的漏洞。 |
| **統計模式逼近** | 概率密度估計 (Probability Density Estimation) | 精準。它是在找資料背後的機率規律。 |
| **難以企及的規律** | 湧現現象 (Emergent Properties) | 精準。模型大到一定程度會出現人類意料之外的能力。 |
| **無法處理全新範式** | "分佈外數據 (Out-of-Distribution, OOD)" | 精準。這是目前 AI 邁向「通用人工智慧 (AGI)」最大的阻礙。 |
